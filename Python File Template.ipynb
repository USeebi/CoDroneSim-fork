## This is an example of what a drone command python file looks like

```python
from codrone_simulator import Drone # 1. Contains all the Drone Commands that moves the Drone

drone = Drone()  # 2. This line sets up your drone in the code so you can start controlling it
drone.pair()     # 3. This line connects your code to the actual drone, getting it ready to follow your commands.

# 4. Under this line Contians the Commands for your drones, you can write any commands you want the drone to follow.
drone.takeoff()
drone.hover(1)
drone.set_roll(0)
drone.set_pitch(50)
drone.set_yaw(0)
drone.set_throttle(0)
drone.move(1)
drone.hover(1)
drone.land()


drone.close() # 5. The drone.close() command safely ends the connection with the drone. It makes sure everything is properly shut down
```
