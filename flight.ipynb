```python
from codrone_simulator import Drone
from codrone_simulator import Note
import time
```

```python
drone = Drone()
drone.pair() # Controller will pair with Drone in Simulator

drone.start_drone_buzzer(Note.C4)
drone.takeoff()
drone.hover(2)
```

```python
# Drone goes up for 1 second with 50 power
drone.set_throttle(50)
drone.move(1)

drone.set_throttle(-50)
drone.move(1)
```

```python
drone.move_forward(distance=100)
time.sleep(2)
drone.move_backward(distance=100)
time.sleep(2)
```

```python

# Drone goes forward for 1 second with 50 power
drone.set_pitch(50)
drone.move(1)

drone.set_pitch(-50)
drone.move(1)
```

```python

drone.move_left(distance=100)
time.sleep(1)
drone.move_right(distance=100)
time.sleep(1)
```

```python

drone.turn_left() # make a default 90 degree left rotate with a 3 second duration
drone.turn_right()
```

```python
drone.set_yaw(10)
drone.move(1)

drone.set_yaw(-10)
drone.move(1)
```

```python
# Drone goes right for 1 second with 10 power
drone.set_roll(10)
drone.move(1)
```

```python
drone.land()
drone.end_drone_buzzer()
drone.close()
```
